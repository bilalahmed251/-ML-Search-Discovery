## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# I would predict whether a page is declining. The label comes from the observed outcome `trend_direction`: pages with `trend_direction == "down"` are labeled as declining. This is an observed label, not a manually invented rule.
import pandas as pd

# Placeholder: You need to load your actual data into a DataFrame named 'df'.
# For demonstration purposes, a dummy DataFrame is created.
data = {
    'trend_direction': ['down', 'up', 'stable', 'down', 'up', 'down', 'up', 'stable'],
    'page_id': [1, 2, 3, 4, 5, 6, 7, 8]
}
df = pd.DataFrame(data)

df["is_declining_label"] = (
    df["trend_direction"]
    .astype(str)
    .str.lower()
    .eq("down")
    .astype(int)
)

print(df["is_declining_label"].value_counts())
print("Declining rate:", round(df["is_declining_label"].mean(), 3))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# My success metric is Precision@50. A good result means that a high proportion of the top 50 pages ranked for refresh are actually declining. This metric is suitable because the SEO team has limited time and needs a reliable shortlist of pages to review first.
def precision_at_k(scores, labels, k=50):
    result = pd.DataFrame({
        "score": scores,
        "label": labels
    }).sort_values("score", ascending=False).head(k)

    return result["label"].mean()

print("Success metric: Precision@50")
```
